Q1.Create two NumPy arrays representing the daily step counts of two friends over a week and use element-wise addition, subtraction, multiplication, and division to compare their activity levels.



```python
import numpy as np
friend1 = np.array([1500, 2500, 3500, 4500, 5500, 6500, 7500, 8500, 9500])
friend2 = np.array([1000, 2000, 3000, 4000, 5000, 6000, 7000, 8000, 9000])

print(" Element-wise Addition:")
print(friend1 + friend2)

print("\nElement-wise Subtraction:")
print(friend1 - friend2)

print("\nElement-wise Multiplication:")
print(friend1 * friend2)

print("\n Element-wise Division:")
print(friend1 / friend2)
```

Q2. Simulate a Spotify-like 'Recommended Songs' feature: Given two 3x3 matrices representing user-song interaction scores (user preferences and song popularity), use dot() and matmul() to compute the final recommendation matrix and explain the difference between the two results.



```python
user_preferences = np.array([
    [5, 4, 3],
    [2, 5, 3],
    [4, 2, 5]
])
song_popularity = np.array([
    [4, 2, 3],
    [3, 5, 2],
    [2, 4, 5]
])
recommendation_dot = np.dot(user_preferences, song_popularity)
recommendation_matmul = np.matmul(user_preferences, song_popularity)

print("\n Recommendation Matrix using dot():")
print(recommendation_dot)

print("\n- Recommendation Matrix using matmul():")
print(recommendation_matmul)

print("\nDifference:")
print("For two 2D matrices, np.dot() and np.matmul() produce the same matrix multiplication result.")
```

Q3. Given a 4x4 NumPy matrix representing the pixel brightness of a small Instagram image, use transpose (T) to rotate the image and then calculate the mean, median, standard deviation, and variance of the pixel values.


```python
pixels = np.array([
    [120, 150, 180, 200],
    [100, 130, 160, 190],
    [90, 140, 170, 210],
    [110, 125, 155, 195]
])

transposed_pixels = pixels.T

print("\nOriginal Pixel Matrix:")
print(pixels)

print("\nTransposed Pixel Matrix:")
print(transposed_pixels)

print("\nMean:", np.mean(pixels))
print("Median:", np.median(pixels))
print("Standard Deviation:", np.std(pixels))
print("Variance:", np.var(pixels))
```

Q4. Take a 3x3 NumPy matrix representing a Zomato restaurant rating correlation grid and use np.linalg.inv(), np.linalg.det(), and np.linalg.eig() to compute its inverse, determinant, and eigenvalues/eigenvectors.<br><br><em><strong>Hint:</strong> If the matrix is not invertible, modify one value and try again.</em>



```python
correlation_grid = np.array([
    [4, 1, 2],
    [1, 5, 1],
    [2, 1, 6]
], dtype=float)

inverse_matrix = np.linalg.inv(correlation_grid)
determinant = np.linalg.det(correlation_grid)
eigenvalues, eigenvectors = np.linalg.eig(correlation_grid)

print("\nOriginal Matrix:")
print(correlation_grid)

print("\nInverse:")
print(inverse_matrix)

print("\nDeterminant:")
print(determinant)

print("\nEigenvalues:")
print(eigenvalues)

print("\nEigenvectors:")
print(eigenvectors)
```

Q5. Create a NumPy array of shape (2, 6) representing the number of orders placed on Swiggy in two cities over 6 days. Reshape it to (3, 4), flatten it, split it into two equal parts, and then stack both parts vertically.



```python
orders = np.array([
    [120, 150, 180, 200, 220, 250],
    [100, 130, 160, 190, 210, 240]
])

print("\nOriginal Array:")
print(orders)

reshaped_orders = orders.reshape(3, 4)

print("\nReshaped to (3, 4):")
print(reshaped_orders)

flattened_orders = orders.flatten()

print("\nFlattened Array:")
print(flattened_orders)

part1, part2 = np.split(flattened_orders, 2)

print("\nFirst Half:")
print(part1)

print("\nSecond Half:")
print(part2)

stacked_orders = np.vstack((part1, part2))

print("\nVertically Stacked:")
print(stacked_orders)
```
