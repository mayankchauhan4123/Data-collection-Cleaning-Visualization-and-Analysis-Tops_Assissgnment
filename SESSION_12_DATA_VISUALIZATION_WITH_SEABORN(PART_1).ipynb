Q1. Install Seaborn in your Python environment and use it to create a histplot showing the distribution of delivery times (in minutes) for 50 Zomato food orders. Generate random data if needed.


```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(42)
delivery_times = np.random.randint(15, 61, 50)

sns.histplot(delivery_times, bins=10, kde=True)

plt.title("Distribution of Zomato Delivery Times")
plt.xlabel("Delivery Time (minutes)")
plt.ylabel("Number of Orders")

plt.show()
```

Q2. Load the 'tips' dataset from Seaborn and create a boxplot to visualize the distribution of total bill amounts by day of the week, similar to how Zomato might analyze spending patterns across weekdays.


```python
import seaborn as sns
import matplotlib.pyplot as plt
tips = sns.load_dataset("tips")
sns.boxplot(data=tips, x="day", y="total_bill")

plt.title("Distribution of Total Bill Amount by Day")
plt.xlabel("Day of the Week")
plt.ylabel("Total Bill Amount")

plt.show()
```

Q3. Simulate IPL match scores for 8 teams and use a violinplot to compare the run distributions per team. Style your plot using the 'darkgrid' Seaborn theme.


```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
np.random.seed(42)
teams = ["CSK", "MI", "RCB", "KKR", "SRH", "RR", "DC", "PBKS"]
data = []

for team in teams:
    scores = np.random.randint(120, 221, 30)

    for score in scores:
        data.append({
            "Team": team,
            "Runs": score
        })

ipl_df = pd.DataFrame(data)

sns.set_theme(style="darkgrid")

plt.figure(figsize=(10, 6))
sns.violinplot(data=ipl_df, x="Team", y="Runs")

plt.title("IPL Run Distribution by Team")
plt.xlabel("IPL Team")
plt.ylabel("Runs")

plt.show()
```

Q4. Create a countplot that displays the number of songs per genre from a list of 40 Spotify tracks, with at least 4 different genres represented.<br><br><em><strong>Hint:</strong> Use a Python list or pandas DataFrame to store your data, then plot with Seaborn.</em>


```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
genres = [
    "Pop", "Rock", "Hip-Hop", "EDM",
    "Pop", "Rock", "Jazz", "Hip-Hop",
    "EDM", "Pop", "Rock", "Jazz",
    "Hip-Hop", "EDM", "Pop", "Rock",
    "Jazz", "Hip-Hop", "EDM", "Pop",
    "Rock", "Jazz", "Hip-Hop", "EDM",
    "Pop", "Rock", "Jazz", "Hip-Hop",
    "EDM", "Pop", "Rock", "Jazz",
    "Hip-Hop", "EDM", "Pop", "Rock",
    "Jazz", "Hip-Hop", "EDM", "Pop"
]
spotify_df = pd.DataFrame({
    "Genre": genres
})

sns.countplot(data=spotify_df, x="Genre")

plt.title("Number of Spotify Songs per Genre")
plt.xlabel("Genre")
plt.ylabel("Number of Songs")

plt.show()
```

Q5.Use Seaborn's kdeplot to visualize the distribution of daily step counts for a week, as if analyzing data from a fitness app. Apply the 'whitegrid' theme and customize the plot color.


```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

steps = [6500, 7200, 8100, 5900, 9000, 10500, 7800]

days = [
    "Monday", "Tuesday", "Wednesday",
    "Thursday", "Friday", "Saturday", "Sunday"
]

steps_df = pd.DataFrame({
    "Day": days,
    "Steps": steps
})

sns.set_theme(style="whitegrid")

sns.kdeplot(
    data=steps_df,
    x="Steps",
    fill=True,
    color="green"
)

plt.title("Distribution of Daily Step Counts")
plt.xlabel("Daily Steps")
plt.ylabel("Density")

plt.show()
```
