Q1. Use plt.subplots() to create a 2x2 grid of subplots and plot four different types of charts (line, bar, scatter, and pie) using any sample data of your choice.


```python
import matplotlib.pyplot as plt
days = ["Mon", "Tue", "Wed", "Thu", "Fri"]
sales = [20, 35, 30, 45, 40]
products = ["A", "B", "C", "D"]
quantities = [25, 40, 30, 50]
heights = [150, 160, 165, 170, 175, 180]
weights = [50, 55, 60, 65, 70, 75]
expenses = [40, 25, 20, 15]
categories = ["Food", "Travel", "Shopping", "Other"]
fig, axes = plt.subplots(2, 2, figsize=(10, 8))
axes[0, 0].plot(days, sales, marker="o")
axes[0, 0].set_title("Line Chart - Weekly Sales")
axes[0, 0].set_xlabel("Day")
axes[0, 0].set_ylabel("Sales")
axes[0, 1].bar(products, quantities, color="orange")
axes[0, 1].set_title("Bar Chart - Product Sales")
axes[0, 1].set_xlabel("Product")
axes[0, 1].set_ylabel("Quantity")
axes[1, 0].scatter(heights, weights, color="green")
axes[1, 0].set_title("Scatter Plot - Height vs Weight")
axes[1, 0].set_xlabel("Height (cm)")
axes[1, 0].set_ylabel("Weight (kg)")
axes[1, 1].pie(expenses, labels=categories, autopct="%1.1f%%")
axes[1, 1].set_title("Pie Chart - Expenses")
plt.tight_layout()
plt.show()
```

Q2.Plot a comparison of average delivery times for Zomato, Swiggy, and Domino's using a bar chart, and style each bar with a different color, linestyle, and linewidth using Matplotlib plot styling options.


```python
import matplotlib.pyplot as plt
platforms = ["Zomato", "Swiggy", "Domino's"]
delivery_time = [35, 40, 30]
bars = plt.bar(platforms, delivery_time)

bars[0].set_color("red")
bars[0].set_edgecolor("black")
bars[0].set_linewidth(2)

bars[1].set_color("orange")
bars[1].set_edgecolor("blue")
bars[1].set_linewidth(3)

bars[2].set_color("blue")
bars[2].set_edgecolor("green")
bars[2].set_linewidth(4)

plt.xlabel("Food Delivery Platform")
plt.ylabel("Average Delivery Time (Minutes)")
plt.title("Comparison of Average Delivery Times")

plt.show()
```

Q3. Create a multi-axis chart that shows the number of Instagram followers (left y-axis) and average daily posts (right y-axis) for five influencers. Use different colors and markers for each axis.


```python
import matplotlib.pyplot as plt

influencers = ["Influencer A", "Influencer B", "Influencer C",
               "Influencer D", "Influencer E"]

followers = [120000, 250000, 180000, 320000, 200000]
daily_posts = [2, 4, 3, 5, 2.5]

fig, ax1 = plt.subplots(figsize=(10, 6))

ax1.plot(
    influencers,
    followers,
    color="blue",
    marker="o",
    linewidth=2,
    label="Followers"
)

ax1.set_xlabel("Influencers")
ax1.set_ylabel("Instagram Followers", color="blue")
ax1.tick_params(axis="y", labelcolor="blue")

ax2 = ax1.twinx()

ax2.plot(
    influencers,
    daily_posts,
    color="red",
    marker="s",
    linewidth=2,
    label="Daily Posts"
)

ax2.set_ylabel("Average Daily Posts", color="red")
ax2.tick_params(axis="y", labelcolor="red")

plt.title("Instagram Followers vs Average Daily Posts")

plt.xticks(rotation=20)

plt.tight_layout()
plt.show()
```

Q4. Plot the number of tickets sold for five recent Bollywood movies (categorical) versus their IMDB ratings (numeric) using a scatter plot. Add annotations to display the movie names above each point.


```python
import matplotlib.pyplot as plt

movies = ["Jawan", "Pathaan", "Animal", "Dunki", "Fighter"]

tickets_sold = [150, 130, 120, 90, 80]

imdb_ratings = [6.9, 5.8, 6.1, 6.4, 6.1]

plt.scatter(tickets_sold, imdb_ratings, color="blue", s=100)

for i in range(len(movies)):
    plt.annotate(
        movies[i],
        (tickets_sold[i], imdb_ratings[i]),
        xytext=(5, 8),
        textcoords="offset points"
    )

plt.xlabel("Tickets Sold (Lakhs)")
plt.ylabel("IMDb Rating")
plt.title("Bollywood Movies: Tickets Sold vs IMDb Ratings")

plt.grid(True, linestyle="--", alpha=0.5)
plt.show()
```

Q5. Add a custom text annotation to a Matplotlib chart showing Flipkart's monthly sales, marking the highest sales point with the label 'Big Billion Days'.<br><br><em><strong>Hint:</strong> Use the ax.annotate() function to place the label at the correct data point.</em>


```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

sales = [120, 135, 150, 145, 160, 175,
         190, 180, 210, 250, 220, 195]

fig, ax = plt.subplots(figsize=(10, 6))

ax.plot(months, sales, marker="o", linewidth=2)

max_sales = max(sales)
max_index = sales.index(max_sales)

ax.annotate(
    "Big Billion Days",
    xy=(months[max_index], max_sales),
    xytext=(20, 20),
    textcoords="offset points",
    arrowprops=dict(arrowstyle="->", linewidth=2),
    fontsize=12
)

ax.set_xlabel("Month")
ax.set_ylabel("Sales (₹ Lakhs)")
ax.set_title("Flipkart Monthly Sales")

ax.grid(True, linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```
