Q1. Install the pandas-profiling library and generate a profile report for the 'Spotify Top 100 Songs' dataset (CSV available on Kaggle); open the HTML report and note any missing values or data type issues.


```python
from google.colab import drive
drive.mount('/content/drive')
import pandas as pd
from ydata_profiling import ProfileReport
file_path = "/content/drive/MyDrive/tops/Spotify 2010 - 2019 Top 100.csv"
df = pd.read_csv(file_path)
profile = ProfileReport(
    df,
    title="Spotify Top 100 Songs - Profile Report",
    explorative=True
)
report_path = "/content/drive/MyDrive/spotify_top_100_profile.html"
profile.to_file(report_path)

print("Data types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isna().sum())

print("\nMissing-value percentages:")
print((df.isna().mean() * 100).round(2))


print("\nDataset shape:")
print(df.shape)

print("\nDuplicate rows:")
print(df.duplicated().sum())

print("\nProfile report saved at:")
print(report_path)
```

Q2. Use Sweetviz to create a comparison report between two CSV files: one containing Zomato restaurant data for Mumbai and another for Delhi. Briefly describe one key difference you spot in the visual report.<br><br><em><strong>Hint:</strong> Use analyze() for single file and compare() for two datasets.</em>


```python
from google.colab import drive
drive.mount('/content/drive')

import pandas as pd
import sweetviz as sv

mumbai_path = "/content/drive/MyDrive/tops/Zomato_Mumbai.csv"
delhi_path = "/content/drive/MyDrive/tops/Zomato_Delhi.csv"

mumbai = pd.read_csv(mumbai_path)
delhi = pd.read_csv(delhi_path)

comparison_report = sv.compare(
    [mumbai, "Mumbai"],
    [delhi, "Delhi"],
    target_feat=None
)

report_path = "/content/drive/MyDrive/zomato_mumbai_vs_delhi.html"

comparison_report.show_html(
    report_path,
    open_browser=False
)

print("Comparison report created successfully!")
print("Report saved at:", report_path)

print("\nMumbai dataset shape:", mumbai.shape)
print("Delhi dataset shape:", delhi.shape)

print("\nMumbai missing values:")
print(mumbai.isna().sum())

print("\nDelhi missing values:")
print(delhi.isna().sum())
```
